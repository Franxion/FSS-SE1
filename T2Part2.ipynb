```python
import git
import os
import itertools
import json
from collections import defaultdict
from tqdm import tqdm

react_repo_path = os.path.join(os.getcwd(), 'react')
if not os.path.exists(react_repo_path):
    print("Cloning the React repository...")
    !git clone https://github.com/facebook/react.git react

task2data_path = os.path.join(os.getcwd(), 'Task2Data')
if not os.path.exists(task2data_path):
    os.makedirs(task2data_path)
```

```python
repo = git.Repo(react_repo_path)
os.path.join(os.getcwd(), 'react')
```

```python
commits = list(repo.iter_commits())
```

```python
pair_counts = defaultdict(int)

print("Processing commits to extract file pairs...")
for commit in tqdm(commits):
    files_changed = list(commit.stats.files.keys())
    if len(files_changed) > 1:
        for file1, file2 in itertools.combinations(sorted(files_changed), 2):
            pair_counts[(file1, file2)] += 1
```

```python
MIN_PAIR_COUNT = 10

frequent_pairs = {pair: count for pair, count in pair_counts.items() if count >= MIN_PAIR_COUNT}

print(f"Total file pairs committed together more than {MIN_PAIR_COUNT} times: {len(frequent_pairs)}")
```

```python
sorted_pairs = sorted(frequent_pairs.items(), key=lambda item: item[1], reverse=True)
top3_pairs = sorted_pairs[:3]

results = []
for ((file1, file2), count) in top3_pairs:
    results.append({
        'file1': file1,
        'file2': file2,
        'degree': count
    })
```

```python
output_file = os.path.join(task2data_path, 'logical_coupling_results(part2).json')

with open(output_file, 'w') as f:
    json.dump({'top3_logical_coupling_pairs': results}, f, indent=4)

print(f"Results saved to {output_file}")
```

```python
print("Top 3 file pairs with the highest degree of logical coupling:")
for pair in results:
    print(f"File 1: {pair['file1']}, File 2: {pair['file2']}, Degree: {pair['degree']}")
```
