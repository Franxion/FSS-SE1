```python
# Filename for the old dependencies
dependencies_old_file = '../part4/dependencies_old(part4).json'

# Run Madge to analyze dependencies at the checked-out commit
!madge --extensions js,jsx --json ./packages > {dependencies_old_file}

print(f"Dependencies at commit {commit_hash} saved to '{dependencies_old_file}'")
```

```python
import json

# Load the dependencies from both JSON files
with open(dependencies_latest_file, 'r') as f:
    dependencies_latest = json.load(f)

with open(dependencies_old_file, 'r') as f:
    dependencies_old = json.load(f)

# Function to compare dependencies
def compare_dependencies(deps_new, deps_old):
    new_deps = {}
    removed_deps = {}

    all_files = set(deps_new.keys()).union(set(deps_old.keys()))

    for file in all_files:
        deps_in_new = set(deps_new.get(file, []))
        deps_in_old = set(deps_old.get(file, []))

        added = deps_in_new - deps_in_old
        removed = deps_in_old - deps_in_new

        if added:
            new_deps[file] = list(added)
        if removed:
            removed_deps[file] = list(removed)

    return new_deps, removed_deps

# Compare dependencies
new_dependencies, removed_dependencies = compare_dependencies(dependencies_latest, dependencies_old)
```

```python
# Filenames for documenting changes
new_dependencies_file = '../part4/new_dependencies(part4).json'
removed_dependencies_file = '../part4/removed_dependencies(part4).json'

# Save the new dependencies introduced to a file
with open(new_dependencies_file, 'w') as f:
    json.dump(new_dependencies, f, indent=2)

# Save the dependencies removed to a file
with open(removed_dependencies_file, 'w') as f:
    json.dump(removed_dependencies, f, indent=2)

print(f"New dependencies introduced have been saved to '{new_dependencies_file}'")
print(f"Dependencies removed have been saved to '{removed_dependencies_file}'")
```
